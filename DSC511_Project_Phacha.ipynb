```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Use your actual dataset
working_df = df_filtered1.copy()

# Define a clean style
plt.style.use('seaborn-v0_8-whitegrid')

# Helper function to plot and annotate with percentages
def plot_categorical_counts_with_pct(data, column_name, title, color_p, orientation='v'):
    vc = data[column_name].value_counts()
    vp = (data[column_name].value_counts(normalize=True) * 100).round(1)
    plot_data = pd.DataFrame({'counts': vc, 'percentage': vp})

    figsize = (12, 6) if orientation == 'v' else (12, 8)
    fig, ax = plt.subplots(figsize=figsize)

    if orientation == 'v':
        ax = plot_data['counts'].plot(kind='bar', color=color_p, ax=ax)
        ax.set_title(f"{title}\nCounts and Percentages", fontsize=16)
        ax.set_xlabel(column_name, fontsize=12)
        ax.set_ylabel("Number of Loans", fontsize=12)
        ax.set_xticks(range(len(plot_data)))
        ax.set_xticklabels(vc.index, rotation=45, ha='right')

        for idx, rect in enumerate(ax.patches):
            height = rect.get_height()
            pct_val = plot_data['percentage'].iloc[idx]
            if height > 0:
                ax.annotate(f'{pct_val}%',
                            xy=(rect.get_x() + rect.get_width() / 2, height),
                            xytext=(0, 5), textcoords="offset points",
                            ha='center', va='bottom', fontsize=10, fontweight='bold', color='black')
    else:
        plot_data = plot_data.sort_values(by='counts', ascending=True)
        ax = plot_data['counts'].plot(kind='barh', color=color_p, ax=ax)
        ax.set_title(f"{title}\nCounts and Percentages", fontsize=16)
        ax.set_xlabel("Number of Loans", fontsize=12)
        ax.set_ylabel(column_name, fontsize=12)
        ax.set_yticklabels(plot_data.index)

        for idx, rect in enumerate(ax.patches):
            width = rect.get_width()
            pct_val = plot_data['percentage'].iloc[idx]
            if width > 0:
                ax.annotate(f'{pct_val}%',
                            xy=(width, rect.get_y() + rect.get_height() / 2),
                            xytext=(5, 0), textcoords="offset points",
                            ha='left', va='center', fontsize=10, fontweight='bold', color='black')

    plt.tight_layout()
    plt.show()

# ==============================================================================
# Generate the 6 Charts
# ==============================================================================

# 1. CHART: BusinessAge
print("\n[CHART 1/6] ORDINAL IMBALANCE CHECK: BusinessAge")
plot_categorical_counts_with_pct(working_df, 'BusinessAge', 'Distribution of Business Age (Ordinal)', 'pink', orientation='v')

# 2. CHART: LoanStatus
print("\n[CHART 2/6] TRADITIONAL TARGET IMBALANCE: LoanStatus")
plot_categorical_counts_with_pct(working_df, 'LoanStatus', 'Distribution of Loan Outcomes (Target)', 'red', orientation='v')

# 3. CHART: BorrState (Top 10)
print("\n[CHART 3/6] UNEVEN GEOGRAPHIC DISTRIBUTION: Top 10 BorrState")
top10_states_vc = working_df['BorrState'].value_counts().head(10)
top10_states_df = working_df[working_df['BorrState'].isin(top10_states_vc.index)]
plot_categorical_counts_with_pct(top10_states_df, 'BorrState', 'Concentration of Top 10 Borrower States', 'teal', orientation='h')

# 4. CHART: BusinessType
print("\n[CHART 4/6] STRUCTURAL IMBALANCE: BusinessType")
plot_categorical_counts_with_pct(working_df, 'BusinessType', 'Distribution of Business Organizational Structure', 'blue', orientation='v')

# 5. CHART: Program
print("\n[CHART 5/6] PROGRAM CONCENTRATION: Specific SBA Programs")
plot_categorical_counts_with_pct(working_df, 'Program', 'Concentration of Specific SBA Programs', 'green', orientation='v')

# 6. CHART: NaicsDescription (Top 5 + Long Tail)
print("\n[CHART 6/6] ECONOMIC SECTOR CONCENTRATION: Top Industry Sectors")
top5_naics_vc = working_df['NaicsDescription'].value_counts().head(5)
naics_plot_series = working_df['NaicsDescription'].copy()
naics_plot_series[~naics_plot_series.isin(top5_naics_vc.index)] = 'OTHER INDUSTRIES/LONG TAIL'
working_df['NaicsPlotBucket'] = naics_plot_series
plot_categorical_counts_with_pct(working_df, 'NaicsPlotBucket', 'Economic Sectors (Top 5 vs Long Tail)', 'red', orientation='h')
```

```python
from sklearn.preprocessing import OrdinalEncoder
import numpy as np
import pandas as pd

def build_machine_learning_matrix(df):
    df_clean = df.copy()

    # 1. ENGINEER FEATURES
    df_clean['LocalLender'] = (df_clean['BorrState'] == df_clean['BankState']).astype(int)

    # Feature 2:
    # census_regions4 = {
        #'Northeast': ['CT', 'ME', 'MA', 'NH', 'RI', 'VT', 'NJ', 'NY', 'PA'],
        #'Midwest': ['IL', 'IN', 'MI', 'OH', 'WI', 'IA', 'KS', 'MN', 'MO', 'NE', 'ND', 'SD'],
        #'South': ['DE', 'FL', 'GA', 'MD', 'NC', 'SC', 'VA', 'DC', 'WV', 'AL', 'KY', 'MS', 'TN', 'AR', 'LA', 'OK', 'TX'],
        #'West': ['AZ', 'CO', 'ID', 'MT', 'NV', 'NM', 'UT', 'WY', 'AK', 'CA', 'HI', 'OR', 'WA']
    #}
    #state_to_region = {state: region for region, states in census_regions.items() for state in states}

    census_regions = {
        'Northeast': ['CT', 'ME', 'MA', 'NH', 'RI', 'VT', 'NJ', 'NY', 'PA'],
        'Midwest': ['IL', 'IN', 'MI', 'OH', 'WI', 'IA', 'KS', 'MN', 'MO', 'NE', 'ND', 'SD'],
        'Southeast': ['AL', 'AR', 'DE', 'FL', 'GA', 'KY', 'LA', 'MD', 'MS', 'NC', 'SC', 'TN', 'VA', 'DC', 'WV'],
        'Southwest': ['AZ', 'NM', 'OK', 'TX'],
        'West': ['AK', 'CA', 'CO', 'HI', 'ID', 'MT', 'NV', 'OR', 'UT', 'WA', 'WY']
    }
    state_to_region = {state: region for region, states in census_regions.items() for state in states}
    df_clean['BorrRegion'] = df_clean['BorrState'].map(state_to_region).fillna('Other')

    try:
        ruca_df = pd.read_csv('/content/drive/MyDrive/RUCA-codes-2020-zipcode.csv')[['ZIPCode', 'PrimaryRUCA']]
        def map_ruca(code):
            if code in [1, 2, 3]: return 'Metropolitan'
            elif code in [4, 5, 6]: return 'Micropolitan'
            elif code in [7, 8, 9]: return 'Small-town'
            elif code == 10: return 'Rural'
            else: return 'Unknown'
        ruca_df['Classification'] = ruca_df['PrimaryRUCA'].apply(map_ruca)
        df_clean = df_clean.merge(ruca_df, left_on='BorrZip', right_on='ZIPCode', how='left')
        df_clean['Classification'] = df_clean['Classification'].fillna('Unknown')
    except FileNotFoundError:
        df_clean['Classification'] = 'Unknown'

    # 2. DROP LEAKAGE, DATES, AND IDENTIFIERS
    cols_to_drop = [
        'AsOfDate', 'ApprovalDate', 'FirstDisbursementDate', 'PaidInFullDate',
        'ChargeOffDate', 'GrossChargeOffAmount', 'BorrName', 'BorrStreet',
        'BorrCity', 'BorrState', 'BorrZip', 'BankName', 'BankStreet', 'BankCity',
        'BankState', 'BankZip', 'LocationID', 'NaicsDescription', 'FranchiseName',
        'FranchiseCode', 'BankFDICNumber', 'BankNCUANumber', 'ProjectCounty', 'ProjectState',
        'SBADistrictOffice', 'CongressionalDistrict', 'SBAGuaranteedApproval', 'LoanStatus',
        'ZIPCode', 'PrimaryRUCA'
    ]
    df_clean = df_clean.drop(columns=[col for col in cols_to_drop if col in df_clean.columns], errors='ignore')

    # 3. DROP REMAINING NaNs
    df_clean = df_clean.dropna()

    # 4. ORDINAL ENCODING
    actual_age_categories = [
        'Unanswered', 'Startup, Loan Funds will Open Business',
        'New Business or 2 years or less', 'Existing or more than 2 years old', 'Change of Ownership'
    ]
    ordinal_encoder = OrdinalEncoder(categories=[actual_age_categories], handle_unknown='use_encoded_value', unknown_value=-1)
    df_clean['BusinessAge_Encoded'] = ordinal_encoder.fit_transform(df_clean[['BusinessAge']])
    df_clean = df_clean.drop(columns=['BusinessAge'])

    # 5. LOG TRANSFORM TARGET
    df_clean['GrossApproval_Log'] = np.log1p(df_clean['GrossApproval'])
    df_clean = df_clean.drop(columns=['GrossApproval'])

    # 6. ONE-HOT ENCODING
    nominal_cols = ['Program', 'ProcessingMethod', 'FixedorVariableInterestInd', 'BusinessType', 'RevolverStatus', 'CollateralInd', 'SoldSecMrktInd', 'BorrRegion', 'Classification']
    df_clean = pd.get_dummies(df_clean, columns=[col for col in nominal_cols if col in df_clean.columns], drop_first=True)

    for col in df_clean.select_dtypes(include=['bool']).columns:
        df_clean[col] = df_clean[col].astype(int)

    return df_clean

# Execute pipeline
preprocessed_df = build_machine_learning_matrix(df_filtered1)

print(f"Matrix shape: {preprocessed_df.shape}")
print(f"Remaining Missing Values: {preprocessed_df.isnull().sum().sum()}")
```

```python
import numpy as np

def machine_ready_diagnostic(df):
    print("=== MACHINE READINESS DIAGNOSTIC ===\n")

    # 1. Check for Missing Values
    missing_count = df.isnull().sum().sum()
    if missing_count == 0:
        print("✅ PASS: 0 missing values.")
    else:
        print(f"❌ FAIL: Found {missing_count} missing values.")

    # 2. Check for Text/Object Data Types
    non_numeric_cols = df.select_dtypes(exclude=[np.number]).columns.tolist()
    if len(non_numeric_cols) == 0:
        print("✅ PASS: All columns are strictly numeric.")
    else:
        print(f"❌ FAIL: Non-numeric columns detected: {non_numeric_cols}")

    # 3. Check for Infinity Values (can happen during log transforms)
    inf_count = np.isinf(df).values.sum()
    if inf_count == 0:
        print("✅ PASS: No infinity values detected.")
    else:
        print(f"❌ FAIL: Found {inf_count} infinity values.")

    # 4. Check for Data Leakage
    leaky_cols = ['GrossApproval', 'SBAGuaranteedApproval', 'LoanStatus']
    found_leaks = [col for col in leaky_cols if col in df.columns]
    if len(found_leaks) == 0:
        print("✅ PASS: Known leakage columns successfully dropped.")
    else:
        print(f"❌ FAIL: Leakage columns still in dataset: {found_leaks}")

    # 5. Check for Target Variable
    if 'GrossApproval_Log' in df.columns:
        print("✅ PASS: Target variable 'GrossApproval_Log' is present.")
    else:
        print("❌ FAIL: Target variable 'GrossApproval_Log' is missing.")

    print("\n====================================")
    if (missing_count == 0) and (len(non_numeric_cols) == 0) and (inf_count == 0) and (len(found_leaks) == 0) and ('GrossApproval_Log' in df.columns):
        print("🚀 VERDICT: 100% READY FOR MACHINE LEARNING.")
    else:
        print("🛑 VERDICT: FIX ERRORS BEFORE MODELING.")

# Run the diagnostic
machine_ready_diagnostic(preprocessed_df)
```
